```python
sentences = ["I love dogs.", "I love cats.", "I love birds."]
```

```python
[sentence.lower().split() for sentence in sentences]
```

```python
def tokenize(text):
    return text.lower().split()

tokenize(sentences[0])
```

```python
def build_vocab(sentences):
    vocab = {}

    for sentence in sentences:
        print("Setence:",sentence)
        tokens = tokenize(sentence)

        for token in tokens:
            if token not in vocab:
                vocab[token] = len(vocab) + 1
                print(f"Token: {token}, ID: {len(vocab)}")
    return vocab

build_vocab(sentences)
```
