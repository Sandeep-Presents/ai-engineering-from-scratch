```python
sentences = [
    "I love my dog",
    "I love my cat",
    "I love Programming",
    "Hi, How are you?",
]

def tokenize(text):
    """
    Tokenizes the input text into a list of tokens.
    Args:
        text (str): The input text to tokenize.
    Returns:
        list: A list of tokens extracted from the input text.
    """
    # Lowercase the text and split by whitespace
    tokens = text.lower().split()
    return tokens
```

```python
tokenize("Hi, How are you?")
```

```python
def build_vocab(sentences):
    """
    Builds a vocabulary from a list of sentences. 
    Args:
        sentences (list): A list of sentences (strings).
    Returns:
        dict: A dictionary mapping each unique token to a unique index.    
    """

    vocab = {}

    for sentence in sentences:
        tokens = tokenize(sentence)
        for token in tokens:
            if token not in vocab:
                vocab[token] = len(vocab) + 1  # Assign a unique index to each token
    return vocab

```
